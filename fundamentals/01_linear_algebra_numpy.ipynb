```python
w = 8.0
learning_rate = 0.1

for step in range(10):
    loss = w**2
    gradient = 2 * w

    w = w - learning_rate* gradient
    
    print("Step:", step, "Weight:", round(w, 4), "Loss:", round(loss, 4), "Gradient:",gradient)
```

```python
import numpy as np
import time

N = 1_000_000

# Plain Python
numbers = list(range(N))

start = time.perf_counter()
python_result = [x * 2 for x in numbers]
python_time = time.perf_counter() - start

# NumPy
arr = np.arange(N)

start = time.perf_counter()
numpy_result = arr * 2
numpy_time = time.perf_counter() - start

print("Python:", python_time, "seconds")
print("NumPy:", numpy_time, "seconds")
print("Speedup:", python_time / numpy_time)
```

```python
import numpy as np

numbers = list(range(1_000_000))
arr = np.arange(1_000_000)

# Confirm equivalent results
assert np.array_equal(np.array(python_result), numpy_result)

# Repeated timing
%timeit [x * 2 for x in numbers]
%timeit arr * 2
```
